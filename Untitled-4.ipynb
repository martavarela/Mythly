```python
import numpy as np
import torch
import matplotlib.pyplot as plt 
import trimesh
import os
import time
import pandas as pd
from IPython.display import display
import scipy.io as sio
import pyvista as pv
import robust_laplacian as rl
from scipy.spatial.distance import cdist

%load_ext autoreload
%autoreload 2

from LibEik3 import EikonalBEMConfig, DataLoader, time_scale_params, scale_time, unscale_time, EikonalGuess, ActivationKinetics, PINN, Trainer, Evaluator
from LibAPDest2 import APDEst
from PlotLib2 import EikonalBEMPlotting

from pyvista.trame.jupyter import launch_server
await launch_server().ready

device = "cuda" if torch.cuda.is_available() else "mps" if torch.backends.mps.is_available() else "cpu"
print(f"Using device: {device}")
```

```python
# --- 1. CONFIGURATION ---
config = EikonalBEMConfig()

# Set up paths
dataset = 'Utah'
ind = 'pvp'
foldername = dataset + str(ind)# Add manually as needed
pathfiles = '../../Jul2026/EDGAR/' + dataset
output_path = '../EikResults/' + dataset + '/'
if os.path.exists(output_path) == False:
    os.makedirs(output_path)

# Read the name of the folder from a CSV file
seed_index = 1
max_seed = 5 # Define the maximum allowed seed index

# Adjust initial weights, decay, and learning scheme limits here.
# reg_data_weight/reg_eikonal_weight are auto-calibrated ONCE from gradient norms at the
# start of training (config.auto_data_weight=True) and then stay fixed; only the ini-loss
# anchor decays over config.epochs (config.auto_ini_decay=True).
config.epochs = 10000

# If >1, train this many independent random seeds and keep whichever reaches the
# lowest final loss -- a simple, effective guard against getting stuck in a poor
# local minimum. Each restart costs a full extra training run, so keep this small.
n_restarts = 1
config.n_restarts = n_restarts

# --- Hard bounds on the initial-guess search windows (ATmin/ATmax are now ABSOLUTE
# times in ms, i.e. directly comparable to tsav, not offsets counted from t1) ---

# If True, find AT_init from the biggest raw-signal peak (either polarity) instead
# of the steepest signal derivative; this is then passed to the APDest fitting
# routine (fit_kinetics_params) as the AT used to personalise the kinetics waveform.
peakAPD = True
config.peakAPD = peakAPD

# Absolute time (ms) where the initial AT guess search window starts (e.g. to skip
# an initial artifact). The AT search window is [ATmin, ATmax].
if dataset == 'Utah':
    ATmin = 0
else:
    ATmin = 200

# Absolute time (ms) where the initial AT guess search window ends: AT_init is
# clipped to [ATmin, ATmax]. Also anchors where the APD/RT search window below
# starts and ends: [ATmax + APDmin, ATmax + APDmax] -- so ATmax + APDmin must
# leave room before t2, or the RT search window is empty.

if dataset == 'Utah':
    ATmax = 100
else:
    ATmax = 350
config.at_max = ATmax
config.at_min = ATmin

# Floor for the initial APD estimate (ms), used to seed rho_kinetics for the PINN
APDmin = 80

# Ceiling for the initial APD estimate (ms): caps the repolarization search window so a
# later beat's QRS isn't mistaken for the T-wave (does NOT bound the PINN's APD output).
APDmax = 180

# Single source of truth for the initialisation/kinetics-fitting APD range, used by both
# EikonalGuess.compute_initial_guess and APDEst.fit_kinetics_params -- kept separate from
# config.rho_bound below, which only bounds the PINN's own learned APD output.
config.init_apd_bound = (APDmin, APDmax)

# Ceiling for the PINN's learned APD output (ms), soft-bounded separately from the
# initialisation search window above -- APDmax caps where RT_init is searched for, while
# APDmaxPINNs caps what APD value the PINN itself is allowed to predict.
APDmaxPINNs = APDmax

# Multiplies the PINN (eikonal) and data losses relative to the ini-loss anchor
loss_factor = 0.01
config.loss_scale = loss_factor

# Sample index splitting the NCC data loss into two windows ([:N] and [N:]); None uses the whole t1-t2 window
config.data_split_idx = None

# Manually biases the auto-calibrated PDE-vs-data balance: >1 favors the eikonal/physics
# term more, <1 favors the data-fitting term more. Independent of loss_factor above,
# which scales PDE+Data *together* relative to ini_loss, not relative to each other.
# factor=1.0 means PDE and data start on genuinely EQUAL gradient footing (this is what
# "balanced" means here) -- the pre-calibration gradient-norm ratio can itself already
# span ~1e15, so huge factors (e.g. 1e12) just saturate the internal [1e-8, 1e18] clamp
# and stop having any effect; keep this in a modest range (e.g. 0.1-100) to meaningfully
# shift the balance. Trainer.train now prints the actual calibrated weights/clamp status.
config.ini_factor = 1
config.end_factor = 200
eikonal_vs_data_factor = 100.0 # np.mean([config.ini_factor, config.end_factor]) # for n_restarts = 1
# print(eikonal_vs_data_factor)
config.eikonal_vs_data_factor = eikonal_vs_data_factor

# Controls how fast the ini-loss anchor decays: it reaches `ini_decay_target_pct` of its
# starting weight after `ini_decay_target_frac` of total epochs. Increase either one to
# make the anchor decay more slowly (linger longer before physics/data take over).
ini_decay_target_frac = 0.1
ini_decay_target_pct = 0.01
config.ini_decay_target_frac = ini_decay_target_frac
config.ini_decay_target_pct = ini_decay_target_pct

# PINN's learned APD output is soft-bounded to this physiological range (ms)
config.rho_bound = (APDmin, APDmaxPINNs)

# Learn activation time only: the network outputs T alone (out_dim=1) and APD/rho
# is NOT estimated per node. compute_loss then falls back to the single global
# kinetics.rho. Set back to True to restore per-node APD estimation.
config.estimate_RT = True

# Choose how to set the Vest kinetics parameters (rho_kinetics, beta1, beta2, beta3):
# - 'default': use the preset/data-driven values as-is, no personalisation
# - 'csv':     reuse a previous run's fitted values from csvresults for this
#              foldername if present, otherwise fall back to fitting fresh (like 'fit')
# - 'fit':     always run the quick A-operator sweep to personalise them fresh
vest_param_mode = 'fit'
config.vest_param_mode = vest_param_mode

# Apply user-defined file paths
config.read_csv = True
csvresults = output_path + 'Results1.csv'
csvname = output_path + 'SIMsdata.csv'

# Suffix for output filenames (PDF, VTK, checkpoint) to distinguish runs with different seeds
output_suffix = f'_seed{seed_index}'

# Set seed for reproducibility
np.random.seed(seed_index)
torch.random.manual_seed(seed_index)
   
print(f"Using foldername: {foldername} with seed index: {seed_index}")

config.t1 = 0
# The recording (data_23.mat) is 1001 samples long (0-1000 ms); t2 is extended
# to 1000 so ATmax + APDmin/APDmax has room to search for repolarization.
config.t2 = 1000
t1 = config.t1

t2 = config.t2
savename = foldername
```

```python
# --- 2. LOAD DATA ---
def struct_field(struct, field):
    # squeeze_me leaves a 0-d object array wrapping the real array; unwrap it
    value = struct[field]
    return value.item() if value.ndim == 0 else value

loader = DataLoader(config)

# Load the data file based on the index
print(f'Loading data from {pathfiles}data_{ind}.mat')
data = sio.loadmat(f'{pathfiles}/data_{ind}.mat')

# Raw recordings have the opposite polarity to the BEM operators, so phie and phie_GT are both negated here
phie = - struct_field(data, 'phie')
elecpos = struct_field(data, 'elec_pos')

u_meas = phie[:, t1:t2]  # Assuming the spikes mean a repeat... (in ms)
tsav = np.arange(phie.shape[1])[t1:t2] # Assuming the spikes mean a repeat... (in ms)

if dataset == 'Halifax' or dataset == 'Bratislava':
    phie_GT = None
    elec_GT = None
    cent = struct_field(data, 'cent')
elif dataset == 'Auckland':
    cent = struct_field(data, 'cent')
    phie_GT = - struct_field(data, 'phieGT')
    elec_GT = struct_field(data, 'GTpos')
elif dataset == 'Utah':
    cent = None  # no GT pacing site for Utah
    phie_GT = - struct_field(data, 'phieGT')
    elec_GT = struct_field(data, 'GTpos')
else:
    raise ValueError(f"Unknown dataset: {dataset}")

# Load the meshes
if dataset == 'Utah':
    mesh = pv.read(f'{pathfiles}/HeartMeshHR.vtk')
    torso = pv.read(f'{pathfiles}/TorsoMeshHR.vtk')
else:
    mesh = pv.read(f'{pathfiles}/HeartMeshHR_{ind}.vtk')
    torso = pv.read(f'{pathfiles}/TorsoMeshHR_{ind}.vtk')
heart_data = mesh.copy()

print(f"Loaded phie data shape: {phie.shape}")
print(f"Loaded electrode positions shape: {elecpos.shape}")
print(f"Loaded measured data shape: {u_meas.shape}")
print(f"Loaded time array shape: {tsav.shape}")

plt.figure(figsize=(10, 4))
pltind = 12
plt.plot(tsav, phie[pltind, t1:t2], label=f'Electrode {pltind}')
plt.xlabel('Time (ms)')
plt.ylabel('Signal (AU)')
plt.show()
```

```python
# --- 3. SPATIAL & TIME SCALING ---
Vpos = np.array(heart_data.points)
faces = np.array(heart_data.faces).reshape(-1, 4)[:, 1:]

print(f"Shape of Vpos: {Vpos.shape}")

# t_centre/t_scale map the [t1, t2] observation window to the PINN's internal ~[0, 1] range
t_centre, t_scale = time_scale_params(config)
t_array_scaled = scale_time(tsav, t_centre, t_scale)

# Spatial coordinates are kept in physical mm units (unscaled) so the PINN
# stays numerically consistent with the precomputed BEM forward operator.
xyz_train = Vpos.copy()

tri_mesh = trimesh.Trimesh(vertices=xyz_train, faces=faces, process=False)
norms = tri_mesh.vertex_normals # are already normalized
norms_tensor = torch.from_numpy(norms).float().to(device)
```

```python
# --- 4. PRECALCULATE BEM OPERATOR ---
# Use physical (unscaled) coordinates so that the Green's function distances and
# Laplacian are in mm, giving the operator the correct physical interpretation.
if dataset == 'Utah':
    A_operator = np.load(f"{pathfiles}/A_operator_small.npy")
else:
    A_operator = np.load(f"{pathfiles}/A_operator_small_{ind}.npy")

A_operator_tensor = torch.from_numpy(A_operator).float().to(device)
print(f"A_operator shape: {A_operator_tensor.shape}")

# Note: Can also load the full operator to test PINNs solver on additional points later
```

```python
# --- 5. INITIAL EIKONAL GUESS ---
# Estimated from the measured electrode signals (u_meas) only; phie_GT is never used here.

AT_init, c0_physical, APD_init, AT_elec, RT_elec = EikonalGuess.compute_initial_guess(
    u_meas, tsav, elecpos, Vpos, config, heart_data)

EikonalBEMPlotting.plot_histograms_and_stats(AT_init, name="AT_init")
if APD_init is not None:
    EikonalBEMPlotting.plot_histograms_and_stats(APD_init, name="APD_init")
    
# Use median of the estimated durations as the global initial guess for `rho_kinetics`
APD_init_median = np.median(APD_init, axis=0)

# Seed rho_kinetics from the data (median initial APD): a rho longer than the observation
# window means Vest never repolarises inside it, so the NCC loss carries no information on rho.
config.rho_kinetics = float(APD_init_median)
print(f"Data-driven rho_kinetics set to: {config.rho_kinetics:.1f} ms  (window: {config.t2 - config.t1} ms)")

# scale everything to the PINN's internal units
AT_init_scaled = scale_time(AT_init, t_centre, t_scale)
AT_init_tensor = torch.from_numpy(AT_init_scaled).float().view(-1, 1).to(device)

# Since we are keeping physical mm coordinates (xyz_train = Vpos), gradT is ∂T/∂x_phys,
# so no spatial scaling factor is needed for c0. We only scale c0 by t_scale because
# the time outputs (AT_pred) will be compressed into the [0, 1] range!
c0_scaled = c0_physical * t_scale

if config.initialrho_fromdata and APD_init_median is not None:
    # Use robust duration scaling (no offset)
    APD_init_scaled = APD_init / t_scale
    APD_init_tensor = torch.from_numpy(APD_init_scaled).float().view(-1, 1).to(device)
    init_target_tensor = torch.cat([AT_init_tensor, APD_init_tensor], dim=1)
else:
    init_target_tensor = AT_init_tensor

EikonalBEMPlotting.plot_initial_guess(heart_data, AT_init, AT_elec, elecpos=elecpos)
EikonalBEMPlotting.plot_initial_guess(heart_data, APD_init, RT_elec - AT_elec, elecpos=elecpos)

EikonalBEMPlotting.plot_electrodes_with_markers(u_meas, tsav, AT_elec, RT_elec)
```

```python
# --- 5b. QUICK KINETICS PARAMETER FIT (optional) ---
# Roughly personalise the activation-kinetics parameters (beta1, beta2, beta3) to
# this recording via a cheap coordinate-wise grid search against the already-computed
# A operator -- no NN training involved. rho_kinetics is never fit; see config.vest_param_mode.
preset_rho, preset_beta1, preset_beta2, preset_beta3 = config.rho_kinetics, config.beta1, config.beta2, config.beta3

print(f"vest_param_mode = '{config.vest_param_mode}'")
print(f"Preset kinetics parameters: rho={preset_rho:.1f} ms, beta1={preset_beta1:.4f} 1/ms, beta2={preset_beta2:.4f} 1/ms, beta3={preset_beta3:.4f} 1/ms")

# 'csv' mode: reuse a previous run's fitted values from csvresults (if present for
# this foldername) instead of re-running the sweep every time.
loaded_from_csv = False
fit_cols = ['Fitted rho_kinetics (ms)', 'Fitted beta1', 'Fitted beta2', 'Fitted beta3', 'Fit NCC Score']
if config.vest_param_mode == 'csv' and os.path.exists(csvresults):
    prev_results = pd.read_csv(csvresults)
    if 'Name' in prev_results.columns and all(col in prev_results.columns for col in fit_cols):
        matches = prev_results[prev_results['Name'] == foldername].dropna(subset=fit_cols[:-1])
        if not matches.empty:
            row = matches.iloc[-1]  # most recent matching run
            fitted_rho, fitted_beta1, fitted_beta2, fitted_beta3 = (
                float(row['Fitted rho_kinetics (ms)']), float(row['Fitted beta1']),
                float(row['Fitted beta2']), float(row['Fitted beta3'])
            )
            # 'Fit NCC Score' is stored as the literal string "N/A" (not real NaN) when a
            # prior run had fitting disabled, so guard the parse rather than using pd.isna.
            try:
                fit_score = float(row['Fit NCC Score'])
            except (TypeError, ValueError):
                fit_score = None
            config.rho_kinetics = fitted_rho
            config.beta1 = fitted_beta1
            config.beta2 = fitted_beta2
            config.beta3 = fitted_beta3
            loaded_from_csv = True
            print(f"Loaded fitted kinetics parameters from {csvresults} for '{foldername}': "
                  f"rho={fitted_rho:.1f} ms, beta1={fitted_beta1:.4f} 1/ms, beta2={fitted_beta2:.4f} 1/ms, beta3={fitted_beta3:.4f} 1/ms")

if config.vest_param_mode == 'default':
    fitted_rho, fitted_beta1, fitted_beta2, fitted_beta3, fit_score = preset_rho, preset_beta1, preset_beta2, preset_beta3, None
    print("vest_param_mode='default': using preset/data-driven kinetics parameters, no personalisation.")
elif loaded_from_csv:
    pass
else:
    # vest_param_mode == 'fit', or 'csv' with no matching row found
    print("Personalising kinetics parameters (rho/beta1/beta2/beta3) via multi-restart Adam+LBFGS...")
    # norm_mode/loss_mode explicitly pinned to this notebook's original behaviour
    # (z-score normalisation + MSE loss), since LibAPDest2.fit_kinetics_params's
    # defaults changed to min-max/MSE (see EikSolverNEW6 for the 3-option comparison).
    fitted_rho, fitted_beta1, fitted_beta2, fitted_beta3, fit_score = APDEst.fit_kinetics_params(
        AT_init, APD_init, tsav, u_meas, A_operator, config, norm_mode='zscore', loss_mode='ncc', device=device
    )
    config.rho_kinetics = fitted_rho
    config.beta1 = fitted_beta1
    config.beta2 = fitted_beta2
    config.beta3 = fitted_beta3
```

```python
# --- 5c. Vest WAVEFORM COMPARISON (preset vs personalised) ---
# Represent the comparison at the earliest-activation node (pacing-site estimate).
example_node = int(np.argmin(AT_init))
AT_node = np.array([AT_init[example_node]])
t_plot = tsav.astype(float)

Vest_preset = APDEst.compute_vest(AT_node, preset_rho, t_plot, preset_beta1, preset_beta2, preset_beta3)[0]
Vest_personalised = APDEst.compute_vest(AT_node, fitted_rho, t_plot, fitted_beta1, fitted_beta2, fitted_beta3)[0]

peak_preset, apd90_preset = APDEst.Calc_CellAPD(Vest_preset, t_plot, apd_min=APDmin, apd_max=APDmax)
peak_pers, apd90_pers = APDEst.Calc_CellAPD(Vest_personalised, t_plot, apd_min=APDmin, apd_max=APDmax)

plt.figure(figsize=(10, 4))
plt.plot(t_plot, Vest_preset, label=(
    f'Preset (rho={preset_rho:.0f}, b1={preset_beta1:.2f}, b2={preset_beta2:.3f}, b3={preset_beta3:.4f})\n'
    f'Peak={peak_preset:.2f}, APD90={apd90_preset:.1f} ms'
))
plt.plot(t_plot, Vest_personalised, label=(
    f'Personalised (rho={fitted_rho:.0f}, b1={fitted_beta1:.2f}, b2={fitted_beta2:.3f}, b3={fitted_beta3:.4f})\n'
    f'Peak={peak_pers:.2f}, APD90={apd90_pers:.1f} ms'
))
plt.xlabel('Time (ms)')
plt.ylabel('Vest (AU)')
plt.title(f'Cellular activation waveform at node {example_node} (AT={AT_node[0]:.1f} ms)')
plt.legend(fontsize=8)
plt.grid()
plt.show()


```

```python
# --- 5d. EXAMPLE ELECTRODE SIGNAL (personalised parameters vs measured) ---
# Assume every heart node follows the same personalised Vest curve, then project
# through A_operator to get the simulated electrode potentials u_sim.
u_sim_personalised = APDEst.simulate_u(AT_init, fitted_rho, t_plot, fitted_beta1, fitted_beta2, fitted_beta3, A_operator)

# Global amplitude scaling (least-squares match), same convention used later for phie_est
scale_factor = np.sum(u_sim_personalised * u_meas) / np.sum(u_sim_personalised**2)
u_sim_scaled = u_sim_personalised * scale_factor

example_elecs =[86, 124]  # matches `pltind` used earlier for raw-signal inspection
plt.figure(figsize=(10, 4))
for example_elec in example_elecs:
    plt.subplot(len(example_elecs), 1, example_elecs.index(example_elec) + 1)
    plt.plot(tsav, u_meas[example_elec], label='Measured', color='blue')
    plt.plot(tsav, u_sim_scaled[example_elec], label='Simulated (personalised kinetics)', color='orange', linestyle='--')
    plt.xlabel('Time (ms)')
    plt.ylabel('Potential (AU)')
    plt.title(f'Electrode {example_elec}: measured vs. personalised-kinetics simulation')
    plt.legend()
    plt.grid()
    plt.show()

# Output the personalised fitted parameters (rho, beta1, beta2, beta3)
print(f'Personalised fitted parameters: rho={fitted_rho:.1f} ms, beta1={fitted_beta1:.4f} 1/ms, beta2={fitted_beta2:.4f} 1/ms, beta3={fitted_beta3:.4f} 1/ms')
```

```python
# --- 6. NEURAL NETWORK TRAINING ---
kinetics = ActivationKinetics(config) # unscaled
print(f"Kinetics used in training: rho={kinetics.rho:.1f} ms, beta1={kinetics.beta1:.4f}, beta2={kinetics.beta2:.4f}, beta3={kinetics.beta3:.4f}")
out_dim = 2 if config.estimate_RT else 1

# Keeping raw physical mm coordinates in the forward pass/physics (no spatial
# scaling). `input_scale` only rescales fc_in's *initial* weights so the huge
# (hundreds-of-mm) raw coordinates don't saturate the output soft-bound's
# sigmoid at init, which otherwise pins every prediction at its boundary and
# gives exactly-zero gradients everywhere.
input_scale = float(np.std(Vpos))
# at_bound/rho_bound confine the PINN's AT and APD outputs to the same windows as the initial guesses
pinn = PINN(
    out_dim=out_dim, 
    hidden_dim=config.hidden_dim, 
    t_scale=t_scale, 
    t_centre=t_centre,
    rho_kinetics=config.rho_kinetics,
    t_bound_margin=config.t_bound_margin,
    rho_bound=config.rho_bound,
    input_scale=input_scale,
    at_bound=(ATmin, ATmax)
)
pinn.to(device)

best_lr = None
# Use configurable final checkpoint name template
if not hasattr(config, 'final_checkpoint_template'):
    config.final_checkpoint_template = 'pinn_checkpoint_t{t1}_to_{t2}.pth'
    
final_checkpoint_path = output_path + config.final_checkpoint_template.format(t1=config.t1, t2=config.t2).replace('.pth', output_suffix + '.pth')

if config.load_checkpoint and hasattr(config, 'checkpoint_path') and os.path.exists(config.checkpoint_path):
    checkpoint = torch.load(config.checkpoint_path, map_location=device)
    try:
        if isinstance(checkpoint, dict) and 'model_state_dict' in checkpoint:
            pinn.load_state_dict(checkpoint['model_state_dict'])
            best_lr = checkpoint.get('final_lr', None)
        else:
            pinn.load_state_dict(checkpoint)
    except RuntimeError:
        pass

if best_lr is None:
    print("Finding optimal Learning Rate...")
    pinn_dummy = PINN(
        out_dim=out_dim, 
        hidden_dim=config.hidden_dim, 
        t_scale=t_scale, 
        t_centre=t_centre,
        rho_kinetics=config.rho_kinetics,
        t_bound_margin=config.t_bound_margin,
        rho_bound=config.rho_bound,
        input_scale=input_scale,
        at_bound=(ATmin, ATmax)
    ).to(device)
    
    lrs, losses = Trainer.find_learning_rate(pinn_dummy, xyz_train, u_meas, t_array_scaled, norms_tensor, A_operator_tensor, c0_scaled, kinetics, config, min_lr=1e-5, max_lr=10, steps=100)
    if len(losses) < 3:
        # LR finder diverged almost immediately (e.g. too few finite steps to take a
        # gradient of); fall back to a conservative default instead of crashing.
        best_lr = 1e-4
        print(f"LR finder produced only {len(losses)} usable points; falling back to default LR: {best_lr:.3e}")
    else:
        # Skip an initial burn-in window: the first few LR-sweep steps are dominated by
        # noisy initialization transients (the EMA smoothing hasn't caught up yet), and
        # their spurious steep drop can otherwise get picked as "steepest descent",
        # locking the search onto an unusably tiny LR (~1e-6) that never trains.
        burn_in = min(10, max(0, len(losses) - 2))
        loss_grd = np.gradient(losses)
        min_loss_idx = burn_in + np.argmin(loss_grd[burn_in:])
        best_lr = lrs[min_loss_idx]/10 # Already adjusted here
        print(f"Optimal Learning Rate: {best_lr:.3e}")

if config.train_model:
    def build_pinn():
        return PINN(
            out_dim=out_dim,
            hidden_dim=config.hidden_dim,
            t_scale=t_scale,
            t_centre=t_centre,
            rho_kinetics=config.rho_kinetics,
            t_bound_margin=config.t_bound_margin,
            rho_bound=config.rho_bound,
            input_scale=input_scale,
            at_bound=(ATmin, ATmax)
        ).to(device)

    if n_restarts > 1:
        # Sweep eikonal_vs_data_factor (instead of just the seed) across restarts;
        # only the best restart's model/history is kept.
        factors = np.linspace(config.ini_factor, config.end_factor, n_restarts)
        restart_rows = []
        best_score = float('inf')
        best_pinn = None
        best_factor = None

        for i, factor in enumerate(factors):
            config.eikonal_vs_data_factor = float(factor)
            print(f"\n=== Restart {i + 1}/{n_restarts} (eikonal_vs_data_factor={factor:.2f}) ===")
            start_time_pinn = time.time()
            model, loss_hist, metrics, flr = Trainer.train(
                build_pinn(), xyz_train, u_meas, t_array_scaled, norms_tensor, A_operator_tensor, c0_scaled, kinetics, config,
                epochs=config.epochs, lr=best_lr, use_lbfgs=config.use_lbfgs, ini_target_tensor=init_target_tensor, verbose=config.verbose,
                seed=seed_index + i
            )
            print(f"Restart {i + 1}/{n_restarts} training time: {time.time() - start_time_pinn:.2f} seconds")

            _, AT_pred_i, _, _, _, _ = Evaluator.evaluate_model(model, xyz_train, tsav, phie_GT, t_centre, t_scale, config, gt_positions=elec_GT)
            dis_i, _ = Evaluator.dist_pacing_site(AT_pred_i, cent, mesh, Vpos, acceptableTmin=5.0, t_window_start=config.t1)

            # Divide by the actual calibrated weight (not `factor`) so a clamped weight is handled correctly.
            total_i = metrics['total'][-1]
            raw_eikonal_i = metrics['eikonal'][-1] / metrics['w_eikonal']
            restart_rows.append({
                'Restart': i + 1,
                'eikonal_vs_data_factor': round(factor, 2),
                'Final Total Loss': total_i,
                'Raw Eikonal (PDE) Loss': raw_eikonal_i,
                'Final Eikonal Loss': metrics['eikonal'][-1],
                'Final Data Loss': metrics['data'][-1],
                'Distance to GT Pacing Site (mm)': dis_i,
            })

            if raw_eikonal_i < best_score:
                best_score = raw_eikonal_i
                best_factor = factor
                best_pinn, loss_hist_pinn, metrics_pinn, final_lr = model, loss_hist, metrics, flr
            else:
                del model  # only the best restart's model is kept in memory

        pinn = best_pinn
        # Restore config (and the notebook-level variable used later for CSV/PDF logging)
        # to the winning restart's factor, not whichever the sweep loop last set it to.
        config.eikonal_vs_data_factor = float(best_factor)
        eikonal_vs_data_factor = float(best_factor)
        print(f"\nBest restart: eikonal_vs_data_factor={best_factor:.2f}, raw eikonal (PDE) loss={best_score:.4e}")
        display(pd.DataFrame(restart_rows))
    else:
        start_time_pinn = time.time()
        pinn, loss_hist_pinn, metrics_pinn, final_lr = Trainer.train(
            pinn, xyz_train, u_meas, t_array_scaled, norms_tensor, A_operator_tensor, c0_scaled, kinetics, config, 
            epochs=config.epochs, lr=best_lr, use_lbfgs=config.use_lbfgs, ini_target_tensor=init_target_tensor, verbose=config.verbose,
            seed=seed_index
        )
        print(f"Training time: {time.time() - start_time_pinn:.2f} seconds")

    torch.save(pinn.state_dict(), final_checkpoint_path)
    print(f"Final model saved to {final_checkpoint_path}")
```

```python
def calculate_Aheart(heart, elecpos):
    # does not include geometric corrections
    # because there are no Neumann boundary conditions 
    # through the epicardium (near the electrodes)
    # Green's function = 1/(4*pi*r)
    # L is the laplacian matrix of the heart mesh, shape: (N, N)
    # included to extract sources from Vsav
    
    verts = np.array(heart.points)
    faces = np.array(heart.faces).reshape(-1, 4)[:, 1:]
    L, M = rl.mesh_laplacian(verts, faces)
    
    # Calculate the distance from each mesh vertex to each electrode
    distances = cdist(verts, elecpos)  # Shape: (N, N_electrodes)
    distances[distances == 0] = 1e-12  # Avoid division by zero for points exactly at electrode positions
    print(f"Distance matrix shape: {distances.shape}")

    # Calculate the Green's function matrix G for the forward problem
    G_matrix = 1 / (4 * np.pi * distances)  # Shape: (N, N_electrodes)
    # Same sign convention as the BEM A_operator (and the negated data): signal = +(L @ G).T @ V
    A = (L @ G_matrix).T

    return A


def normalise_rows(x):
    # Per-electrode amplitude normalisation (max |value| = 1); the A_heart operator has arbitrary units
    return x / np.maximum(np.max(np.abs(x), axis=1, keepdims=True), 1e-12)


# if measured phieGT exists (Utah, Auckland), estimate these signals and compute their agreement with the measured signals
if phie_GT is not None:
    phieGT = phie_GT[:, t1:t2]
    # estimate operator for points on heart surface
    A_heart = calculate_Aheart(mesh, elec_GT)
    # For every node in heart mesh, estimate action potential using estimated AT map
    phieGT_est_raw = A_heart @ Vest_all

    # Compare normalised waveforms: A_heart has no physical amplitude scale, so a single global
    # least-squares factor collapses the estimate to ~0; RMSE below is therefore a normalised RMSE.
    phieGT_n = normalise_rows(phieGT)
    phieGT_est_n = normalise_rows(phieGT_est_raw)

    # Electrode-specific metrics and their global mean +/- std
    per_elec_rmse_GT = np.sqrt(np.mean((phieGT_est_n - phieGT_n) ** 2, axis=1))
    per_elec_ncc_GT = np.array([np.corrcoef(phieGT_est_n[e], phieGT_n[e])[0, 1] for e in range(phieGT.shape[0])])
    mean_rmse_GT, std_rmse_GT = float(np.nanmean(per_elec_rmse_GT)), float(np.nanstd(per_elec_rmse_GT))
    mean_ncc_GT, std_ncc_GT = float(np.nanmean(per_elec_ncc_GT)), float(np.nanstd(per_elec_ncc_GT))
    print(f"phieGT normalised RMSE: mean={mean_rmse_GT:.4f}, std={std_rmse_GT:.4f} | NCC: mean={mean_ncc_GT:.4f}, std={std_ncc_GT:.4f}")

    EikonalBEMPlotting.plot_electrodes(phieGT_n, phieGT_est_n, tsav, elec_GT, ylabel=r'normalised $\phi_e$')
    EikonalBEMPlotting.plot_rmse_bar(per_elec_rmse_GT, title=f"phieGT per-electrode normalised RMSE: mean={mean_rmse_GT:.3f} \u00b1 {std_rmse_GT:.3f}")
    EikonalBEMPlotting.plot_gof_bar(per_elec_ncc_GT, title=f"phieGT per-electrode NCC: mean={mean_ncc_GT:.3f} \u00b1 {std_ncc_GT:.3f}")
    EikonalBEMPlotting.plot_phie_gt_maps(heart_data, elec_GT, per_elec_ncc_GT, per_elec_rmse_GT, savename=output_path)

    # Add the global metrics to the CSV row and to the table printed in the PDF
    results_df.loc[row_index, 'Mean phieGT RMSE'] = mean_rmse_GT
    results_df.loc[row_index, 'Std phieGT RMSE'] = std_rmse_GT
    results_df.loc[row_index, 'Mean phieGT NCC'] = mean_ncc_GT
    results_df.loc[row_index, 'Std phieGT NCC'] = std_ncc_GT
    df_rmse = df_rmse[~df_rmse['Metric'].str.startswith('phieGT')]
    df_rmse.loc[len(df_rmse)] = ['phieGT normalised RMSE (mean +/- std)', f"{mean_rmse_GT:.4f} +/- {std_rmse_GT:.4f}"]
    df_rmse.loc[len(df_rmse)] = ['phieGT NCC (mean +/- std)', f"{mean_ncc_GT:.4f} +/- {std_ncc_GT:.4f}"]
else:
    per_elec_rmse_GT = per_elec_ncc_GT = None

results_df.to_csv(csvresults, index=False)
display(results_df)
```

```python
# --- 8. FINAL QUALITY METRICS ---
def rng(x):
    return f"{np.nanmin(x):.1f} - {np.nanmax(x):.1f} ms (mean {np.nanmean(x):.1f})"

def mean_std(x, fmt):
    return f"{np.nanmean(x):{fmt}} +/- {np.nanstd(x):{fmt}}"

summary = [
    ('Dataset / run', f"{foldername} (seed {seed_index})"),
    ('Distance to GT pacing site (mm)', f"{dis:.2f}" if dis is not None else "N/A"),
    ('phie NCC (mean +/- std)', mean_std(per_elec_ncc, '.4f')),
    ('phie RMSE (mean +/- std)', mean_std(per_elec_rmse, '.4f')),
    ('phieGT NCC (mean +/- std)', mean_std(per_elec_ncc_GT, '.4f') if per_elec_ncc_GT is not None else "N/A"),
    ('phieGT normalised RMSE (mean +/- std)', mean_std(per_elec_rmse_GT, '.4f') if per_elec_rmse_GT is not None else "N/A"),
    ('Estimated AT range', rng(AT_pred)),
    ('Estimated APD range', rng(APD_pred) if config.estimate_RT else "N/A"),
    ('GT AT range', rng(AT_gt) if AT_gt is not None else "N/A"),
    ('GT APD range', rng(APD_gt) if APD_gt is not None else "N/A"),
    ('AT RMSE vs GT (ms)', f"{rmse_at:.3f}" if AT_gt is not None else "N/A"),
    ('APD RMSE vs GT (ms)', f"{rmse_apd:.3f}" if APD_gt is not None else "N/A"),
    ('Spatial NCC AT', f"{cc_AT:.4f}" if cc_AT is not None else "N/A"),
    ('Spatial NCC APD', f"{cc_APD:.4f}" if cc_APD is not None else "N/A"),
]
df_quality = pd.DataFrame(summary, columns=['Metric', 'Value'])
display(df_quality)
df_quality.to_csv(f'{output_path}{foldername}{output_suffix}_quality_metrics.csv', index=False)
```
